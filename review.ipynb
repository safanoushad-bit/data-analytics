# sum of multiplies
### A finance script needs the sum of all numbers divisible by 3 or 5 between 1 and 50 for a quick 

```python
sum=0
for i in range(1,51):
    if i % 3==0 or i % 5==0:
        sum=sum+i
print("sum of multiplies of 3 or 5",sum)
```
