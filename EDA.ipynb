## univariate analysis

```python
plt.figure(figsize=(8,5))
sns.histplot(df["BMI"])
plt.title("BMI distribution")
plt.xlabel("BMI")
plt.ylabel("Number of patients")
plt.show()
```

## interpretation
The BMI values are approximately normally distributed, with most patients having BMI values in the middle range. Only a few patients have very low or very high BMI values.

## bivariate analysis

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df,x="Gender",y="BMI")
plt.title("Gender vs BMI")
plt.xlabel("Gender")
plt.ylabel("BMI")
plt.show()
```

## interpretation
The box plot shows that the BMI distribution is very similar for both males and females. The median BMI of males is slightly higher than that of females, while both groups have a few outliers with very low and very high BMI values. Overall, there is no significant difference in BMI between males and females.

```python
plt.figure(figsize=(5,4))
corr=df[["BMI","Cholesterol"]].corr()
sns.heatmap(corr,annot=True)
plt.title("correlation between BMI and Cholesterol")
plt.show()
```

## interpretation
The correlation coefficient between BMI and cholesterol is 0.018, indicating almost no linear relationship between the two variables. Changes in BMI do not significantly affect cholesterol levels in this dataset.

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df,x="Gender",y="Cholesterol")
plt.title("Gender vs Cholesterol")
plt.show()
```

## interpretation
The cholesterol levels of males and females are very similar, with only a slight difference in the median values. The spread of cholesterol is also comparable, indicating no significant difference between genders.
